```python
import pandas as pd
from hampel import hampel #this allows the hampel package to run
```

```python
#reading C1 continuous data
C1hobodf = pd.read_csv("P1_creek_current.csv") 
C1hobodf["Date-Time"] = pd.to_datetime(C1hobodf["Date-Time"]) 
C1date = C1hobodf["Date-Time"]
C1hobodfdate = C1date[1:]
```

```python
#filtering the C1 data for the outlier spikes where the hobo probe was removed from the piezometer
C1filteredhobo = hampel(C1hobodf["Temperature"], window_size=6, n_sigma=1.5)
C1filteredhoboseries = pd.Series(C1filteredhobo.filtered_data, index = C1hobodf.index)
C1filteredhobo = C1filteredhoboseries[1:]
C1filteredhobo
```
