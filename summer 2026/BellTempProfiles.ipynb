```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import matplotlib.dates as mdates
import scipy
```

```python
temps = pd.read_csv('BellDepthProfiles.csv')
```

```python
temps.head()
```

```python
fig, (ax1, ax2, ax3, ax4) = plt.subplots(1,4, figsize=(8,8))
ax1.plot(temps.P1tempBefore, temps.decimalfeet, label = 'P1 before pumping')
ax1.plot(temps.P1TempAfter, temps.decimalfeet, label = 'P1 after pumping')
ax2.plot(temps.P2TempBefore, temps.decimalfeet, label = 'P2 before pumping')
ax2.plot(temps.P2TempAfter, temps.decimalfeet, label = 'P2 after pumping')
ax3.plot(temps.P5Temp, temps.decimalfeet, label = 'P5')
ax4.plot(temps.P6Temp, temps.decimalfeet, label = 'P6')
# ax.yaxis.set_inverted(True)
ax1.set_ylim(22, 5)
ax2.set_ylim(22, 5)
ax3.set_ylim(22, 5)
ax4.set_ylim(22, 5)
ax1.legend()
ax2.legend()
ax3.legend()
ax4.legend()
plt.show()
```
