```python
hobodf = pd.read_csv("P1_All_6.16.csv")
hobodf["Date-Time (CST/CDT)"] = pd.to_datetime(hobodf["Date-Time (CST/CDT)"])

date = hobodf["Date-Time (CST/CDT)"]
temp = hobodf["Temperature (W-CT 22414702), °C"]
mandate = pd.to_datetime(["11/11/2025 12:00:00", "11/18/2025 12:00:00", "1/06/2026 12:00:00", "1/13/2026 12:00:00", "2/3/2026 12:00:00", "2/17/2026 12:00:00", "2/24/2026 12:00:00", "3/3/2026 12:00:00", "3/10/2026 12:00:00", "3/31/2026 12:00:00", "4/7/2026 12:00:00", "4/14/2026 12:00:00", "4/21/2026 12:00:00", "4/28/2026 12:00:00", "5/12/2026 12:00:00", "5/19/2026 12:00:00", "5/26/2026 12:00:00", "6/16/2026  12:00:00"])
mantemp = [15, 14.2, 9.2, 8.9, 7.5, 6.7, 6.3, 6, 5.6, 5.4, 5.5, 5.6, 5.8, 7.3, 7, 7.2, 7.8, 9.8]

fig, ax = plt.subplots()
ax.plot(date, temp, color="blue")
ax.plot(mandate, mantemp, 'x', color = "red")
plt.xlabel("Year-Month")
plt.ylabel("Temperature (°C)")
plt.title("Continuous P1 temperature")
plt.legend

plt.show()
```

```python
df = pd.read_csv("P1_Month_MayOnly.csv")
df["Date-Time (CST/CDT)"] = pd.to_datetime(df["Date-Time (CST/CDT)"])

may_time = df["Date-Time (CST/CDT)"]
may_temp = df["Temperature (W-CT 22414702), °C"]

plt.figure(figsize=(10, 5))
plt.plot(may_time, may_temp, color="blue")
plt.xlabel("Month-Day")
plt.ylabel("Temperature (°C)")
plt.title("Continuous P1 temperature in May")

plt.show()
```
