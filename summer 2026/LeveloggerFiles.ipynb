```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
bridge = pd.read_csv("BridgeLevellogger.6.23.26.csv", header = 11, index_col = ["DateTime"])
bridge["DateTime"] = pd.to_datetime(bridge.index)
bridge
```

```python
fig, ax = plt.subplots(layout = 'constrained')
ax.scatter(bridge.index, bridge.LEVEL, s = 1, color = "teal", label = "water level at the bridge")
plt.ylim(0.1, 0.4)

# ax.locator_params(axis = "x", nbins=4)
plt.xticks(rotation = 45)

plt.show()
```
